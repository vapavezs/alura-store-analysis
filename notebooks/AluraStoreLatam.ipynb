#1. Análisis de facturación



En este primer análisis se calcula el ingreso total de cada tienda sumando los valores de la columna Precio.

```python
ingreso_tienda1 = tienda1['Precio'].sum()
ingreso_tienda2 = tienda2['Precio'].sum()
ingreso_tienda3 = tienda3['Precio'].sum()
ingreso_tienda4 = tienda4['Precio'].sum()

print("Ingreso total Tienda 1:", ingreso_tienda1)
print("Ingreso total Tienda 2:", ingreso_tienda2)
print("Ingreso total Tienda 3:", ingreso_tienda3)
print("Ingreso total Tienda 4:", ingreso_tienda4)
```

```python
ingresos = pd.DataFrame({
    'Tienda': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'Ingreso Total': [
        ingreso_tienda1,
        ingreso_tienda2,
        ingreso_tienda3,
        ingreso_tienda4
    ]
})

ingresos
```

A partir del cálculo del ingreso total, se observa que existen diferencias relevantes entre las tiendas. Esta métrica será clave para identificar el rendimiento general de cada una en los siguientes análisis.

# Informe final



##Recomendación de tienda a vender

El objetivo de este análisis fue evaluar el desempeño de las cuatro tiendas de la cadena Alura Store, con el fin de identificar cuál de ellas presenta el menor rendimiento general y, por lo tanto, debería ser vendida para permitir al Sr. Juan iniciar un nuevo emprendimiento.

Para ello, se analizaron distintos indicadores clave como ingresos totales, ventas por categoría, calificaciones de los clientes, productos más y menos vendidos y el costo de envío promedio.

## Análisis de resultados

En primer lugar, el análisis de los ingresos totales mostró diferencias significativas entre las tiendas. Mientras algunas presentan un mayor volumen de ventas, otras generan ingresos considerablemente menores, lo que evidencia un desempeño económico más bajo.

Respecto a las categorías de productos, se observó que ciertas tiendas concentran sus ventas en pocas categorías, mientras que otras presentan una distribución más equilibrada. Las tiendas con menor diversidad de categorías tienden a depender de un número reducido de productos, lo que puede representar un riesgo comercial.

El análisis de las calificaciones promedio de los clientes permitió evaluar el nivel de satisfacción. Algunas tiendas obtienen evaluaciones consistentemente más bajas, lo que sugiere posibles problemas en la calidad del servicio o de los productos ofrecidos.

Al identificar los productos más y menos vendidos, se detectó que en algunas tiendas existen productos con muy baja rotación, lo que puede indicar una gestión ineficiente del catálogo.

Finalmente, el costo de envío promedio también mostró diferencias relevantes. Dado que este costo es asumido por el cliente, un valor elevado puede afectar negativamente la experiencia de compra y reducir la competitividad de la tienda.


## Conclusión y recomendación final

Tras analizar el desempeño de las cuatro tiendas de la cadena Alura Store, se concluye que la Tienda 4 es la opción más adecuada para ser vendida. Esta tienda presenta el menor ingreso total, debilidades en categorías relevantes, una calificación promedio que no destaca frente a las demás y una baja rotación de productos estratégicos, incluso considerando que ofrece el menor costo de envío.

En comparación, tiendas como la Tienda 3 muestran un mejor equilibrio entre volumen de ventas, satisfacción del cliente y desempeño por categoría, lo que evidencia una mayor eficiencia operativa. Por estas razones, se recomienda que el Sr. Juan venda la Tienda 4 para iniciar su nuevo emprendimiento.


#Extra: Análisis del desempeño geográfico

```python
tienda1[['lat', 'lon']].head()
```

```python
tienda1['Tienda'] = 'Tienda 1'
tienda2['Tienda'] = 'Tienda 2'
tienda3['Tienda'] = 'Tienda 3'
tienda4['Tienda'] = 'Tienda 4'
```

```python
df_geo = pd.concat([tienda1, tienda2, tienda3, tienda4])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

for tienda, color in zip(
    ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    ['blue', 'green', 'orange', 'red']
):
    data = df_geo[df_geo['Tienda'] == tienda]
    plt.scatter(data['lon'], data['lat'], alpha=0.3, label=tienda)

plt.xlabel('Longitud')
plt.ylabel('Latitud')
plt.title('Distribución geográfica de las ventas por tienda')
plt.legend()
plt.show()
```

Como análisis adicional, se exploró la distribución geográfica de las ventas utilizando las coordenadas de latitud y longitud. A partir de un gráfico de dispersión, se observó que las cuatro tiendas presentan patrones geográficos similares, concentrando sus ventas en zonas comparables.

No se identificaron diferencias significativas en la cobertura geográfica que expliquen el menor desempeño de la Tienda 4. Esto sugiere que su bajo rendimiento no está relacionado con la ubicación de las ventas, sino con factores comerciales como el volumen de ventas, la rotación de productos y los ingresos generados.
